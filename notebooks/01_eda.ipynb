# Premier League Match Outcome Prediction - EDA

Exploration of Premier League match data from the 2016/17 season to 2025/26

```python
# imports

import pandas as pd
from pathlib import Path
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
DATA_DIR = Path("../data/raw")
files = sorted(DATA_DIR.glob("E0_*.csv"))

print(f"found {len(files)} files")
```

```python
# combine data

dfs = []

for file in files:
    df_season = pd.read_csv(file)
    
    season = file.stem.replace("E0_", "")
    df_season["Season"] = season
    
    dfs.append(df_season)
    
df = pd.concat(dfs, ignore_index = True)
```

```python
print("0% missing:", (missing_pct == 0).sum())
print("0–10% missing:", ((missing_pct > 0) & (missing_pct < 0.10)).sum())
print("10–50% missing:", ((missing_pct >= 0.10) & (missing_pct < 0.50)).sum())
print("50–90% missing:", ((missing_pct >= 0.50) & (missing_pct < 0.90)).sum())
print("90%+ missing:", (missing_pct >= 0.90).sum())
```
